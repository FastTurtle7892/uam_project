```python
import folium
from PIL import Image
import sqlite3
import time

# 데이터베이스 연결 설정 (SQLite 예시)
conn = sqlite3.connect('C:/Users/user/database/gps.db')  # 'gps.db' 파일로 연결 설정
cursor = conn.cursor()

# 운동장의 중심 좌표와 경계 좌표 설정
location = [37.6108075, 126.9963554]
bounds = [
    [37.61087526430053, 126.99549931574487],  # 남서쪽 경계 좌표
    [37.610841866408634, 126.99681784541491]  # 북동쪽 경계 좌표
]

# Folium 지도 객체 생성
map_folium = folium.Map(location=location, zoom_start=17)

# 업로드한 이미지를 사용하여 Folium에 추가
image_path = 'C:/Users/user/database/kookmin.jpg'  # 경로 확인 필요
image_overlay = folium.raster_layers.ImageOverlay(
    image=image_path,
    bounds=bounds
)
image_overlay.add_to(map_folium)

# 지도 업데이트 함수
def update_map(locations):
    # Folium 지도 객체 재생성
    map_folium = folium.Map(location=location, zoom_start=17)
    image_overlay.add_to(map_folium)
    
    # 위치 데이터로 마커 추가
    for lat, lon in locations:
        folium.Marker([lat, lon], popup='Location').add_to(map_folium)
    
    # 지도를 HTML 파일로 저장
    map_folium.save('C:/Users/user/database/map_with_realtime_location.html')  # 경로 확인 필요

# 위치 데이터 가져오기 함수
def fetch_locations():
    cursor.execute("SELECT LAT, LON FROM Signal ORDER BY ROWID DESC LIMIT 10")
    return cursor.fetchall()

# 메인 실행 로직
updates = 10
for _ in range(updates):
    locations = fetch_locations()
    update_map(locations)
    time.sleep(5)

# 데이터베이스 연결 닫기
conn.close()

# 노트북에서 지도를 표시하려면 다음과 같이 합니다.
map_folium
```
